```python
import matplotlib.pyplot as plt

# 1. 准备数据
regions = ["华北", "华南", "东北", "西北", "西南", "华东"]
sales = [2354, 1902, 3524, 2698, 2896, 2563]
total = sum(sales)
ratio = sales[2] / total * 100  # 东北占比

# 2. 设置中文
plt.rcParams['font.sans-serif'] = ['SimHei']  # Windows用SimHei，Mac用Arial Unicode MS
plt.rcParams['axes.unicode_minus'] = False

# 3. 创建画布
fig, ax = plt.subplots(figsize=(12, 8), dpi=120)
fig.patch.set_facecolor('#191f3f')  # 深蓝色背景
ax.set_facecolor('#191f3f')

# 4. 绘制柱状图
bars = ax.bar(regions, sales, color="#00b0f0", alpha=0.88)

# 在柱子上方写数字
for bar in bars:
    h = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2, h + 45, f"{h}", ha="center", va="bottom", color="white")

# 5. 坐标轴与网格
ax.set_ylim(0, 4000)
ax.set_yticks([0,1000,2000,3000,4000])
ax.tick_params(axis='x', colors='white', labelsize=11)
ax.tick_params(axis='y', colors='white', labelsize=11)
ax.grid(axis="y", color="#434a70", linestyle="--", alpha=0.6)
ax.set_axisbelow(True)

# 隐藏边框
for s in ax.spines.values():
    s.set_visible(False)

# 6. 标题
title = f"3月各区域销量分布\n东北销量最多占比总销量的{ratio:.0f}%，华南销量最低"
ax.set_title(title, color="white", loc="left", fontsize=18, pad=20)

# 底部注释
plt.figtext(0.11, 0.07, "*注: 数据来源于公司销售系统，统计日期截至2022.03.31", color="white", fontsize=9)

plt.show()
```

```python
import matplotlib.pyplot as plt

# 数据
regions = ["华北", "华南", "东北", "西北", "西南", "华东"]
sales = [2354, 1902, 3524, 2698, 2896, 2563]
avg_value = sum(sales) / len(sales)

# 中文设置
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 创建画布
fig, ax = plt.subplots(figsize=(12, 8), dpi=120)
fig.patch.set_facecolor('#191f3f')
ax.set_facecolor('#191f3f')

# 绘制柱状图
bars = ax.bar(regions, sales, color="#0070c0")

# 柱子上方标注数值
for bar in bars:
    h = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2, h + 40, f"{h}", ha="center", va="bottom", color="white")

# 绘制平均值黄色横线
ax.axhline(y=avg_value, color='#ffd700', linewidth=2)
ax.text(5.1, avg_value + 40, f"平均值: {avg_value:.0f}", color="#ffd700", fontsize=12)

# 坐标轴设置
ax.set_ylim(0, 4000)
ax.set_yticks([0,1000,2000,3000,4000])
ax.tick_params(axis='x', colors='white', labelsize=11)
ax.tick_params(axis='y', colors='white', labelsize=11)
ax.grid(axis="y", color="#434a70", linestyle="--", alpha=0.6)
ax.set_axisbelow(True)

# 隐藏边框
for s in ax.spines.values():
    s.set_visible(False)

# 标题
total = sum(sales)
ratio = sales[2] / total * 100
title_text = f"3月各区域销量分布\n东北销量最多占比总销量的{ratio:.0f}%，华南销量最低"
ax.set_title(title_text, color="white", loc="left", fontsize=18, pad=20)

# 底部注释
plt.figtext(0.11, 0.07, "*注: 数据来源于公司销售系统，统计日期截至2022.03.31", color="white", fontsize=9)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 数据
quarters = ["2021Q1", "Q2", "Q3", "Q4", "2022Q1", "Q2"]
sales = [3121, 4086, 4321, 4601, 4936, 4231]
profit = [1020, 1421, 1502, 1623, 1781, 1432]

# 中文设置
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 创建画布
fig, ax = plt.subplots(figsize=(12,8), dpi=120)
fig.patch.set_facecolor('#191f3f')
ax.set_facecolor('#191f3f')

# 分组柱状图
x = np.arange(len(quarters))
width = 0.35
bar1 = ax.bar(x - width/2, sales, width, label="销售额", color="#0070c0")
bar2 = ax.bar(x + width/2, profit, width, label="利润额", color="#e04e6c")

# 在柱子上方标注数值
for bar in bar1:
    h = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2, h + 80, f"{h}", ha="center", va="bottom", color="white")
for bar in bar2:
    h = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2, h + 80, f"{h}", ha="center", va="bottom", color="white")

# X轴标签
ax.set_xticks(x)
ax.set_xticklabels(quarters)

# 图例
ax.legend(loc="upper right", facecolor="#191f3f", labelcolor="white")

# Y轴
ax.set_ylim(0, 5500)
ax.tick_params(axis='x', colors='white', labelsize=11)
ax.tick_params(axis='y', colors='white', labelsize=11)
ax.grid(axis="y", color="#434a70", linestyle="--", alpha=0.6)
ax.set_axisbelow(True)

# 隐藏边框
for s in ax.spines.values():
    s.set_visible(False)

# 标题
title_text = "2021年至今季度销售额(万)和利润额(万)\n2022年第二季度销售额首次出现下降，降幅达到15%"
ax.set_title(title_text, color="white", loc="left", fontsize=18, pad=20)

# 底部注释
plt.figtext(0.11, 0.07, "*注: 数据来源于公司销售系统，统计日期截至2022.06.30", color="white", fontsize=9)

plt.show()
```
