```python
import matplotlib.pyplot as plt

# 数据
x_labels = ["5月", "6月", "7月", "8月", "9月", "10月", "11月", "12月", "1月", "2月", "3月"]
sales = [146, 198, 296, 412, 506, 615, 789, 1021, 3782, 3215, 2936]

# 中文设置
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 创建画布
fig, ax = plt.subplots(figsize=(12,8), dpi=120)
fig.patch.set_facecolor('#191f3f')
ax.set_facecolor('#191f3f')

# 绘制折线
ax.plot(x_labels, sales, color='#e06c48', linewidth=2)

# 最高点标注+虚线
peak_idx = 8
peak_val = sales[peak_idx]
ax.vlines(x=peak_idx, ymin=0, ymax=peak_val, color='#e06c48', linestyle='--', alpha=0.7)
ax.text(peak_idx, peak_val + 120, f"{peak_val}", color="white", ha="center", fontsize=12)

# Y轴设置
ax.set_ylim(0, 4000)
ax.set_yticks([0,1000,2000,3000,4000])
ax.tick_params(axis='x', colors='white', labelsize=11)
ax.tick_params(axis='y', colors='#ff8866', labelsize=11)
ax.grid(axis="y", color="#434a70", linestyle="--", alpha=0.6)
ax.set_axisbelow(True)

# 隐藏边框
for s in ax.spines.values():
    s.set_visible(False)

# 底部色块示意
from matplotlib.patches import Rectangle
# 2021色块
rect1 = Rectangle((0.2, -500), 7.3, 250, facecolor="#60c2d8", edgecolor="none")
ax.add_patch(rect1)
ax.text(3.8, -380, "2021", color="white", ha="center", fontsize=12)
# 2022色块
rect2 = Rectangle((7.6, -500), 2.2, 250, facecolor="#f2bc4f", edgecolor="none")
ax.add_patch(rect2)
ax.text(8.7, -380, "2022", color="white", ha="center", fontsize=12)
ax.set_ylim(-800, 4000)

# 标题
title_text = "化妆品品类月度销量走势\n2022年销量迅速增加，1月最高，销量达到3782"
ax.set_title(title_text, color="white", loc="left", fontsize=18, pad=20)

# 底部注释
plt.figtext(0.11, 0.07, "*注: 数据来源于公司销售系统，统计日期截至2022.03.31", color="white", fontsize=9)

plt.show()
```

```python
import matplotlib.pyplot as plt

# 数据
months = ["1月", "2月", "3月", "4月", "5月", "6月", "7月", "8月"]
rate = [0.536, 0.498, 0.527, 0.708, 0.609, 0.496, 0.586, 0.704]

# 中文设置
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 创建画布
fig, ax = plt.subplots(figsize=(12,8), dpi=120)
fig.patch.set_facecolor('#191f3f')
ax.set_facecolor('#191f3f')

# 绘制棒棒糖图：竖线 + 菱形标记
for i, val in enumerate(rate):
    ax.vlines(x=i, ymin=0, ymax=val, color='#e04e6c', linewidth=2)
    ax.plot(i, val, marker='D', color='#e04e6c', markersize=8)
    ax.text(i, val + 0.02, f"{val*100:.2f}%", ha="center", va="bottom", color="white", fontsize=11)

# X轴
ax.set_xticks(range(len(months)))
ax.set_xticklabels(months, color="white", fontsize=11)

# Y轴
ax.set_ylim(0, 0.8)
ax.set_yticks([])
ax.tick_params(axis='x', colors='white')

# 隐藏边框
for s in ax.spines.values():
    s.set_visible(False)
ax.spines["bottom"].set_color("white")

# 标题
title_text = "2022年1-8月公司计划完成率\n公司整体完成率55%，4月和8月超过70%，2月和6月较低未过半"
ax.set_title(title_text, color="white", loc="left", fontsize=18, pad=20)

# 底部注释
plt.figtext(0.11, 0.07, "*注: 数据来源于公司销售系统，统计日期截至2022.08.31", color="white", fontsize=9)

plt.show()
```

```python
import matplotlib.pyplot as plt

# 数据
months = ["1月", "2月", "3月", "4月", "5月", "6月"]
y2021 = [1686, 1345, 1934, 1658, 1865, 1936]
y2022 = [1385, 1846, 1654, 1936, 2564, 2236]

# 中文设置
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 创建画布
fig, ax = plt.subplots(figsize=(12,8), dpi=120)
fig.patch.set_facecolor('#191f3f')
ax.set_facecolor('#191f3f')

# 绘制两条折线，带圆点标记
line1 = ax.plot(months, y2021, color='#e04e6c', marker='o', linewidth=2, label="2021年")
line2 = ax.plot(months, y2022, color='#0070c0', marker='o', linewidth=2, label="2022年")

# 给每个点标注数值
for x, val in enumerate(y2021):
    ax.text(x, val + 60, f"{val}", ha="center", color="white", fontsize=11)
for x, val in enumerate(y2022):
    ax.text(x, val + 60, f"{val}", ha="center", color="white", fontsize=11)

# Y轴
ax.set_ylim(0, 3000)
ax.set_yticks([0,500,1000,1500,2000,2500,3000])
ax.tick_params(axis='x', colors='white', labelsize=11)
ax.tick_params(axis='y', colors='white', labelsize=11)
ax.grid(axis="y", color="#434a70", linestyle="--", alpha=0.6)
ax.set_axisbelow(True)

# 图例
ax.legend(loc="upper right", facecolor="#191f3f", labelcolor="white", frameon=False)

# 隐藏边框
for s in ax.spines.values():
    s.set_visible(False)

# 标题
title_text = "2022年上半年各月同比去年销量\n上半年同比去年增长明显，5月份同比增长最多，增长近40%"
ax.set_title(title_text, color="white", loc="left", fontsize=18, pad=20)

# 底部注释
plt.figtext(0.11, 0.07, "*注: 数据来源于公司销售系统，统计日期截至2022.06.30", color="white", fontsize=9)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 中文设置
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 创建画布
fig, ax = plt.subplots(figsize=(12,8), dpi=120)
fig.patch.set_facecolor('#191f3f')
ax.set_facecolor('#191f3f')

# 数据
percent = 0.65
# 底层完整圆环
ax.pie([1], radius=1, colors=['none'], wedgeprops={'width':0.03, 'edgecolor':'#0070c0'})
# 进度半圆
ax.pie([percent, 1-percent], radius=0.9, startangle=90,
       colors=['#0070c0','#191f3f'],
       wedgeprops={'width':0.7, 'edgecolor':'none'})

# 中间文字
ax.text(0,0,f"{percent*100:.0f}%",ha="center",va="center",color="white",fontsize=45)

# 标题
title_text = "2022年上半年目标完成率\n截至6月30日销售目标总体完成率达到65%"
ax.set_title(title_text, color="white", loc="left", fontsize=20, pad=20)

# 底部注释
plt.figtext(0.11, 0.07, "*注: 数据来源于公司销售系统", color="white", fontsize=10)

ax.axis("equal")
plt.show()
```
