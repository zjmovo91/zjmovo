```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

fig, ax1 = plt.subplots(figsize=(9, 6))
regions = ["华北", "华南", "东北", "西北", "西南", "华东"]
s2022 = [2354, 1902, 3524, 2698, 2896, 2563]
s2021 = [2021, 1563, 3213, 2531, 2631, 2361]
yoy = [16, 22, 10, 7, 10, 9]

x = np.arange(len(regions))
width = 0.35

ax1.bar(x - width / 2, s2021, width, label="2021销量", color="#a9c4eb")
ax1.bar(x + width / 2, s2022, width, label="2022销量", color="#4472c4")
ax1.set_xticks(x)
ax1.set_xticklabels(regions)
ax1.set_ylabel("销量")

ax2 = ax1.twinx()
ax2.plot(x, yoy, marker="o", color="#ed7d31", label="同比去年(%)")
ax2.set_ylabel("同比%")

lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc="upper left")
ax1.set_title("图27 簇状柱形折线图")
plt.tight_layout()
plt.savefig("fig27_簇状柱形折线.png", dpi=150)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

fig, ax = plt.subplots(figsize=(12, 6))
months = [
    "1月", "2月", "3月",
    "4月", "5月", "6月",
    "7月", "8月", "9月",
    "10月", "11月", "12月"
]
month_sale = [2354, 1902, 3524, 2698, 2896, 2563, 3156, 2896, 3621, 2635, 2963, 2789]
quarter_sale = [
    7780, 7780, 7780,
    8157, 8157, 8157,
    9673, 9673, 9673,
    8387, 8387, 8387
]
x = np.arange(len(months))
w = 0.35

ax.bar(x - w / 2, month_sale, w, label="月度销量", color="#4472c4")
ax.bar(x + w / 2, quarter_sale, w, label="季度销量", color="#70ad47")
ax.set_xticks(x)
ax.set_xticklabels(months)
ax.set_title("图28 复合柱形图 月度&季度销量")
ax.legend()
plt.tight_layout()
plt.savefig("fig28_复合柱形图.png", dpi=150)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

fig, ax = plt.subplots(figsize=(8, 5))
regions = ["华东", "西北", "东北", "华北", "华南"]
complete = [35, 51, 62, 74, 86]
y = np.arange(len(regions))

ax.barh(y, [100]*len(regions), height=0.4, color="#e8e8e8")
ax.barh(y, complete, height=0.4, color="#4472c4")

for i, v in enumerate(complete):
    ax.text(v + 2, i, f"{v}%", va="center")

ax.set_yticks(y)
ax.set_yticklabels(regions)
ax.set_xlim(0, 100)
ax.set_title("图29 滑珠图 区域完成率")
ax.set_xlabel("完成率 %")
plt.tight_layout()
plt.savefig("fig29_滑珠图.png", dpi=150)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

fig, ax = plt.subplots(figsize=(8, 5))
regions = ["华东", "西北", "东北", "华北", "华南"]
c2022 = [35, 51, 62, 74, 86]
c2021 = [45, 39, 53, 69, 92]
y = np.arange(len(regions))

ax.barh(y, [100]*len(regions), height=0.4, color="#e8e8e8")
ax.scatter(c2021, y, s=90, color="#ed7d31", zorder=10, label="2021完成率")
ax.scatter(c2022, y, s=90, color="#4472c4", zorder=10, label="2022完成率")

ax.set_yticks(y)
ax.set_yticklabels(regions)
ax.set_xlim(0, 100)
ax.set_title("图30 对比滑珠图 两年完成率对比")
ax.set_xlabel("完成率 %")
ax.legend()
plt.tight_layout()
plt.savefig("fig30_对比滑珠图.png", dpi=150)
plt.show()
```
