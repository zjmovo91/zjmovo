```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Circle, PathPatch
from matplotlib.path import Path

plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

fig, ax = plt.subplots(figsize=(12,8), dpi=120)
fig.patch.set_facecolor('#191f3f')
ax.set_facecolor('#191f3f')
ax.set_xlim(-1.2,1.2)
ax.set_ylim(-1.2,1.2)
ax.axis("off")

# 外圈圆环
outer_circle = Circle((0,0), 1, color="none", ec="#0070c0", lw=2)
ax.add_patch(outer_circle)

percent = 0.65
wave_amp = 0.08
x = np.linspace(-1,1,300)
# 波浪曲线
y_wave = percent - 1 + wave_amp * np.sin(x*12)
# 构建填充路径
verts = []
codes = []
verts.append((-1,-1))
codes.append(Path.MOVETO)
for xi, yi in zip(x, y_wave):
    verts.append((xi, yi))
    codes.append(Path.LINETO)
verts.append((1, -1))
codes.append(Path.LINETO)
verts.append((-1, -1))
codes.append(Path.CLOSEPOLY)

path = Path(verts, codes)
patch = PathPatch(path, facecolor="#0070c0", edgecolor="none")
ax.add_patch(patch)

# 圆形裁剪
clip_circle = Circle((0,0), 1)
patch.set_clip_path(clip_circle)

# 中间文字
ax.text(0,0, f"{percent*100:.0f}%", color="white", ha="center", va="center", fontsize=45)

# 标题
title_text = "本科及以上学历员占比\n6月30日最新统计数据本科及以上员工占比65%"
ax.set_title(title_text, color="white", loc="left", fontsize=20, pad=20)

# 底部注释
plt.figtext(0.11, 0.07, "*注: 数据来源于公司人力资源系统", color="white", fontsize=10)

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 数据
labels = ['[20,30)', '[30,40)', '[40,50)', '>=50']
sizes = [37.5, 29.2, 20.8, 12.5]
colors = ['#e04e6c','#f2bc57','#17a2a3','#3498db']

fig, ax = plt.subplots(figsize=(12,8), dpi=120)
fig.patch.set_facecolor('#191f3f')
ax.set_facecolor('#191f3f')

# 空心圆环图
wedges, texts, autotexts = ax.pie(
    sizes,
    labels=labels,
    colors=colors,
    autopct='%1.1f%%',
    startangle=90,
    pctdistance=0.75,
    labeldistance=1.05,
    wedgeprops={'width':0.2, 'edgecolor':'#191f3f'}
)

# 设置文字颜色为白色
for t in texts:
    t.set_color('white')
    t.set_fontsize(13)
for a in autotexts:
    a.set_color('white')
    a.set_fontsize(12)

# 标题
title_text = "2022年上半年年龄分布\n公司平均年龄32.5，23-30员工比例最高"
ax.set_title(title_text, color="white", loc="left", fontsize=20, pad=20)

# 底部注释
plt.figtext(0.11,0.07,"*注: 数据来源于公司人力资源系统，统计日期截至2022.06.30",color="white",fontsize=10)

ax.axis('equal')
plt.show()
```
