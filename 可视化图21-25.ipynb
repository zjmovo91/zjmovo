```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

fig, ax = plt.subplots(figsize=(8, 4), subplot_kw={"projection": "polar"})
ax.set_theta_direction(-1)
ax.set_theta_offset(np.pi / 2.0)

vmin, vmax = 50, 150
value = 76
theta_min = np.pi
theta_max = 0

def val2ang(v):
    return theta_min - (v - vmin)/(vmax-vmin)*(theta_min-theta_max)

# 背景环
bg_theta = np.linspace(theta_min, theta_max, 200)
bg_r = np.full_like(bg_theta, 0.8)
ax.plot(bg_theta, bg_r, color="#e8e8e8", lw=20)

# 指针
needle_angle = val2ang(value)
ax.plot([needle_angle, needle_angle], [0, 0.9], color="#c82423", lw=3)

# 刻度
tick_vals = [50,60,70,80,90,100,110,120,130,140,150]
tick_angles = [val2ang(v) for v in tick_vals]
ax.set_xticks(tick_angles)
ax.set_xticklabels([str(x) for x in tick_vals])
ax.set_ylim(0, 1)
ax.set_yticks([])
ax.set_title("仪表盘图，指针数值：76", pad=20, fontsize=13)

plt.tight_layout()
plt.savefig("fig22_gauge.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

fig, ax1 = plt.subplots(figsize=(9, 5))

years = [2017, 2018, 2019, 2020, 2021, 2022]
sales = [1603, 2106, 2406, 3265, 3721, 3921]
yoy = [27, 31, 14, 36, 14, 5]

x = np.arange(len(years))
ax1.bar(x, sales, color="#4472c4", width=0.5, label="销售量")
ax1.set_xticks(x)
ax1.set_xticklabels(years)
ax1.set_ylabel("销售量")

ax2 = ax1.twinx()
ax2.plot(x, yoy, color="#ed7d31", marker="o", linestyle="-", label="同比(%)")
ax2.set_ylabel("同比 %")

ax1.set_title("图23 柱形折线图")
fig.legend(loc="upper left", bbox_to_anchor=(0.12,0.92))

plt.tight_layout()
plt.savefig("fig23_bar_line.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

fig, ax = plt.subplots(figsize=(10,6))

goods = ["口红","面膜","隔离","防晒","精华","面霜"]
actual = np.array([653, 523, 648, 856, 714, 785])
target = np.array([700, 500, 600, 900, 600, 600])
pass_val = np.array([600, 600, 600, 600, 600, 600])
good_val = pass_val + 200
exc_val = good_val + 200

y = np.arange(len(goods))
h = 0.6

# 三层背景区间：及格、良好、优秀
for i in range(len(goods)):
    ax.barh(y[i], exc_val[i], height=h, color="#e6e6e6")
    ax.barh(y[i], good_val[i], height=h, color="#c9c9c9")
    ax.barh(y[i], pass_val[i], height=h, color="#a0a0a0")

# 实际值
ax.barh(y, actual, height=h*0.4, color="#2e5fa8", zorder=10)

# 目标红线
for i,t in enumerate(target):
    ax.vlines(t, y[i]-h/2, y[i]+h/2, color="#d92121", lw=2, zorder=11)

ax.set_yticks(y)
ax.set_yticklabels(goods)
ax.set_title("图25 子弹图")

plt.tight_layout()
plt.savefig("fig25_bullet.png", dpi=150, bbox_inches="tight")
plt.show()
```
