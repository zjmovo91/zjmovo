```python
# -*- coding: utf-8 -*-
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams["font.sans-serif"] = ["Noto Sans CJK SC", "SimHei"]
plt.rcParams["axes.unicode_minus"] = False
plt.rcParams["figure.dpi"] = 150

BLUE = "#3b82f6"
BLUE2 = "#60a5fa"
DARK = "#1e293b"

# 数据
bf6 = ["华东", "西北", "东北", "华北", "华南"]
y22 = [1215, 1321, 1426, 1531, 2238]   #2022销量
y21 = [1003, 1265, 1531, 1436, 2066]   #2021销量

fig, ax = plt.subplots(figsize=(9, 5.5))
y = np.arange(len(bf6))[::-1]

ax.barh(y, y21, height=0.45, color=BLUE2, label="2021年销量", zorder=2)
ax.barh(y, [-v for v in y22], height=0.45, color=BLUE, label="2022年销量", zorder=2)

# 数值标签
for yi, v in zip(y, y21):
    ax.text(v + 40, yi, f"{v}", va="center", fontsize=9, color=BLUE2)
for yi, v in zip(y, y22):
    ax.text(-v - 40, yi, f"{v}", va="center", ha="right", fontsize=9, color=BLUE)

ax.axvline(0, color=DARK, lw=1)
ax.set_yticks(y)
ax.set_yticklabels(bf6)
ax.set_xlim(-2600, 2600)
ax.set_title("6.蝴蝶图 —— 区域销量对比", color=DARK, fontsize=11, fontweight="bold")
ax.legend(frameon=False, loc="lower right")

for s in ["top", "right", "bottom"]:
    ax.spines[s].set_visible(False)
ax.tick_params(axis="x", labelbottom=False, bottom=False)

plt.tight_layout()
fig.savefig("chart6_蝴蝶销量.png", bbox_inches="tight", facecolor="white")
plt.show()
```

```python
# -*- coding: utf-8 -*-
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams["font.sans-serif"] = ["Noto Sans CJK SC", "SimHei"]
plt.rcParams["axes.unicode_minus"] = False
plt.rcParams["figure.dpi"] = 150

BLUE2 = "#60a5fa"
ORANGE = "#f59e0b"
RED = "#ef4444"
DARK = "#1e293b"

#数据
goods9 = ["口红", "面膜", "隔离", "防晒", "精华"]
s21 = [3568, 4135, 4436, 4106, 4936]
s22 = [2569, 3241, 2965, 3209, 3541]

fig, ax = plt.subplots(figsize=(9, 5.5))
x = np.arange(len(goods9))
w = 0.36

ax.bar(x - w/2, s21, width=w, color=BLUE2, label="2021销量", zorder=2)
ax.bar(x + w/2, s22, width=w, color=ORANGE, label="2022销量", zorder=2)

for xi, v in zip(x - w/2, s21):
    ax.text(xi, v + 80, f"{v}", ha="center", va="bottom", fontsize=9)
for xi, v in zip(x + w/2, s22):
    ax.text(xi, v + 80, f"{v}", ha="center", va="bottom", fontsize=9)

#差值
for xi, a, b in zip(x, s21, s22):
    ax.text(xi, max(a,b)+260, f"Δ{a-b}", ha="center", va="bottom", fontsize=9,
            color=RED, fontweight="bold")

ax.set_xticks(x)
ax.set_xticklabels(goods9)
ax.set_ylim(0, 6200)
ax.legend(frameon=False, loc="upper left")
ax.set_title("9.对比柱形图 ——2021 vs 2022销量", color=DARK, fontsize=11, fontweight="bold")

for s in ["top", "right"]:
    ax.spines[s].set_visible(False)
ax.grid(axis="y", linestyle="--", alpha=0.3)
ax.set_axisbelow(True)

plt.tight_layout()
fig.savefig("chart9_分组对比柱形.png", bbox_inches="tight", facecolor="white")
plt.show()
```

```python
# -*- coding: utf-8 -*-
import numpy as np
from datetime import datetime, timedelta
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

plt.rcParams["font.sans-serif"] = ["Noto Sans CJK SC", "SimHei"]
plt.rcParams["axes.unicode_minus"] = False
plt.rcParams["figure.dpi"] = 150

BLUE = "#3b82f6"
GRAY_LIGHT = "#e2e8f0"
DARK = "#1e293b"

#任务数据
tasks = [
    ("制定计划", datetime(2022, 3, 1), 11, 0.51),
    ("方案设计", datetime(2022, 3, 13), 8, 0.32),
    ("资源调配", datetime(2022, 3, 22), 10, 0.21),
    ("第一阶段", datetime(2022, 4, 2), 13, 0.85),
    ("第二阶段", datetime(2022, 4, 16), 24, 0.36),
    ("第三阶段", datetime(2022, 5, 11), 14, 0.68),
    ("项目总结", datetime(2022, 5, 26), 7, 0.68),
]

fig, ax = plt.subplots(figsize=(11, 5.5))

for i, (name, start, days, comp) in enumerate(tasks):
    y_pos = len(tasks)-1 - i
    done_days = days * comp
    #总时长底色
    ax.barh(y_pos, days, left=start, height=0.5, color=GRAY_LIGHT, zorder=2)
    #已完成部分
    ax.barh(y_pos, done_days, left=start, height=0.5, color=BLUE, zorder=3)
    ax.text(start + timedelta(days=days), y_pos, f"{comp*100:.0f}%",
            va="center", ha="left", fontsize=9, color=DARK, zorder=4)

yticks_pos = [len(tasks)-1 - i for i in range(len(tasks))]
ax.set_yticks(yticks_pos)
ax.set_yticklabels([t[0] for t in tasks])

#x轴日期格式化
ax.xaxis.set_major_formatter(FuncFormatter(lambda ord_val,_: datetime.fromordinal(int(ord_val)).strftime("%m/%d")))
ax.set_title("10.甘特图 — 项目进度计划", color=DARK, fontsize=11, fontweight="bold")

for s in ["top", "right"]:
    ax.spines[s].set_visible(False)
ax.grid(axis="x", linestyle="--", alpha=0.3)
ax.set_axisbelow(True)

plt.tight_layout()
fig.savefig("chart10_gantt.png", bbox_inches="tight", facecolor="white")
plt.show()
```
